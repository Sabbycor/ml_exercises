## ONE-HOT ENCODING con variabili categoriche

PANDAS

```python
df = pd.get_dummies(df, columns = ["colore"])
df.head()
```

```python
df = pd.get_dummies(df, columns = ["colore"], prefix="color", prefix_sep="-")
df.head()
```

NUMPY

Utilizza un trasformatore di SciKit-Learn

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
X = df.values
X.shape
```

```python
transf = ColumnTransformer(
    [
        ("ohe", OneHotEncoder(), [1])
    ], # la funzione così composta droppa automaticamente le colonne che non sono interessate dalle trasformazioni -> usa remainder per tenerle
    remainder= "passthrough"
)
```

```python
X = transf.fit_transform(X)
X
```
