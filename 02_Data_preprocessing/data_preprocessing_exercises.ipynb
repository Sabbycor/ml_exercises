```python
df_drop = df.copy()
```

```python
tresh = int(df_drop.shape[0]*0.7)
tresh
```

```python
df_drop = df_drop.dropna(axis=1, thresh=354)
df_drop.head(10)
```

```python
df_drop.shape
```

```python
tresh = int(df_drop.shape[1]*0.75)
tresh
```

```python
df_drop = df_drop.dropna(axis=0, thresh=9)
df_drop.head(10)
```

```python
df_drop.shape
```

```python
df_drop = df_drop.dropna(subset = ["PRICE"])
df_drop.head()
```

```python
df_drop.isna().sum()
```

```python
replace_with = df_drop.mean(numeric_only=True)
df_drop = df_drop.fillna(replace_with)
df_drop.head(30)
```

```python
df_drop.isna().sum()
```

```python
df_drop["CRIM"].unique()
```

```python
df_drop["CRIM"] = df_drop["CRIM"].map({"LOW": 1, "MODERATE": 2, "HIGH": 3, "VERY HIGH": 4})
df_drop.head(10)
```

```python
df_drop["CRIM"].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
le.fit(df_drop.CRIM)
df_drop["categorical_label"] = le.transform(df_drop.CRIM)

```

```python
df_drop["CHAS"].unique()
```

```python
le = LabelEncoder()
df_drop["CHAS"] = le.fit_transform(df_drop["CHAS"])
df_drop.head()
```

```python
df_drop["CHAS"].unique()
```

```python
# standardizzazione - vedi a quali colonne applicarla

df_std = df_drop.copy()
```

```python
df_std.describe()
```

```python
from sklearn.preprocessing import StandardScaler

cols_to_exclude = ['PRICE', 'CHAS', 'categorical_label']
cols_to_scale = [c for c in df_std.columns if c not in cols_to_exclude]

scaler = StandardScaler()
df_std[cols_to_scale] = scaler.fit_transform(df_std[cols_to_scale])
```

 standardizzi i numeri che misurano "quanto" (tassi, percentuali, distanze, importi), non quelli che indicano "che categoria" (0/1 o etichette), e mai il target insieme alle feature.

```python
features = ["CRIM", "ZN", "INDUS", "NOX", "RM", "AGE", "DIS", "RAD", "TAX", "PTRATIO", "B"]
```

```python
to_std = df_std[features]


df_std[features] = (to_std - to_std.mean()) / to_std.std()
df_std.head(10)
```

```python
round(df_std[cols_to_scale].describe(), 2)
```

```python
df_std[features].std()
```

```python
df_std[features].mean()
```

```python
df_std.to_csv("housing_cleaned.tsv", sep="\t")
```
