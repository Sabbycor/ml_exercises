```python
import pandas as pd
```

IMPORTARE CSV

```python
df = pd.read_csv(BASE_URL + "shirts_example.csv", index_col=0)
df.head()
```

con index_col si trasforma la prima colonna che dà di default la funzione quando legge i dati e la si trasforma in indice

TSV

```python
df = pd.read_csv(BASE_URL + "shirts_example.tsv", index_col=0, sep= "\t")
df.head()
```

"\t" è il carattere che indica come separatore "|".
Pandas non fa distinzione tra csv e tsv
