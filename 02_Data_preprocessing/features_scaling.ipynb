## Feature Scaling

Portare tutte le feature del nostro dataset all'interno di un unico range di valori.

### Normalizzazione

```python
import numpy as np
import pandas as pd
```

```python
BASE_URL = "https://raw.githubusercontent.com/ProfAI/machine-learning-fondamenti/refs/heads/main/datasets/"
```

```python
wine = pd.read_csv(BASE_URL + "wine.csv", usecols=[0, 1, 7])
wine.head(10)
```

### Pandas

```python
wine_norm = wine.copy()
```

```python
features = ["alcol", "flavonoidi"]
to_norm = wine_norm[features]
wine_norm[features] = (to_norm - to_norm.min()) / (to_norm.max() - to_norm.min())
round(wine_norm.head(10), 2)
```

```python
wine_norm[features].min()
```

```python
wine_norm[features].max()
```

### SciKit_Learn

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
mms = MinMaxScaler()
X = wine.drop("classe", axis=1).values # la feature target non va normalizzata, quindi la rimuoviao dall'array
#axis=0 è per le righe
X_norm = mms.fit_transform(X)
print(X_norm.min(), X_norm.max())
print(X_norm[:5])
```

## Standardizzazione

```python
wine_std = wine.copy()
```

```python
features = ["alcol", "flavonoidi"]
to_std = wine_std[features]


wine_std[features] = (to_std - to_std.mean()) / to_std.std()
wine_std.head(10)
```

```python
# per verificare si calcola il valore medio, che deve essere 0 o tendente allo 0.

wine_std[features].mean()
```

```python
# la deviazione standard dovrebbe essere tendente all'1

wine_std[features].std()
```

### Scikit-Learn

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
X = wine.drop("classe",axis=1).values
X_std = ss.fit_transform(X)
X_std[:5]
```

```python
X_std.mean()
```

```python
X_std.std()
```

i numeri che otteniamo sono leggermente diversi. Questo perchè scikitlearn utilizza numpy per calcolare la deviazione standard. numpy e pandas la calcolano in modi diversi. Se vogliamo uniformità anche con pandas:

```python
wines_std = wine.copy()

features = ["alcol","flavonoidi"]
to_std = wine_std[features]
wine_std[features] = (to_std - to_std.mean())/to_std.std(ddof=0) #ddof=0 per replicare i risultati di sklearn. Aggiugiamo questo parametro e lo settiamo a 0. Normalmente è 1.
wine_std[:5]
```

La differenza è questa:

Deviazione standard della <b> popolazione </b>: usa tutti i dati disponibili della popolazione
Deviazione standard del campione: usa solo un sottoinsieme dei dati e corregge il calcolo per stimare meglio la variabilità della popolazione
La formula cambia nel denominatore:

Popolazione: si divide per N
Campione: si divide per N - 1
Dove N è il numero di osservazioni.

Perché compare N - 1 nel campione?

Perché quando usi un <b> campione</b>, stai cercando di stimare la variabilità dell’intera popolazione partendo da pochi dati. Dividere per N - 1 invece che per N introduce una piccola correzione che rende la stima meno “ottimista” e più accurata in media.

Esempio pratico: supponi di avere i valori [2, 4, 6].

Calcoli la media: 
μ = 2 + 4 + 63 = 4
μ= 32+4+6 = 4

Calcoli gli scarti dalla media:

2 - 4 = -2
4 - 4 = 0
6 - 4 = 2
Sommi i quadrati: 
4 + 0 + 4 = 8
4+0+4=8

Varianza e deviazione standard:

Popolazione: varianza = 8 / 3
Campione: varianza = 8 / 2
Quindi la deviazione standard del campione risulta più grande.

Nel video, la differenza tra Pandas e scikit-learn nasce proprio da questo:

Pandas di default usa la deviazione standard campionaria
NumPy, e quindi spesso anche scikit-learn, usa quella della popolazione
