## Codifica del target tramite label endocing

```python
import pandas as pd
```

```python
BASE_URL = "https://raw.githubusercontent.com/ProfAI/machine-learning-fondamenti/refs/heads/main/datasets/"
```

```python
df = pd.read_csv(BASE_URL + "shirts_sold.csv", index_col=0)
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
x = LabelEncoder() #inizializza
df["venduta"] = x.fit_transform(df["venduta"])
df.head()
```

```python
x.classes_
```

```python
# se vogliamo invece fare una trasformazione inversa (0,1 -> Vero, Falso)

y = [0, 0, 1, 0, 1]

x.inverse_transform(y)
```
