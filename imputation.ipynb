## Eseguire l'imputazione per i dati mancanti

sostituire i valori mancanti con altri valori

```python
import numpy as np
import pandas as pd
```

### MEAN IMPUTATION
sostituire con il valore medio per la colonna

```python
BASE_URL = "https://raw.githubusercontent.com/ProfAI/machine-learning-fondamenti/refs/heads/main/datasets/"
```

```python
iris = pd.read_csv(BASE_URL + "iris_missing.csv")
iris.head()
```

```python
iris_imp = iris.copy()
```

verifichiamo quale colonna ha più valori mancanti:

sepal_width     10

```python
col = "sepal_width"
```

```python
replace_with = round(iris_imp[col].mean(), 1)
replace_with
```

```python
iris_imp[col] = iris_imp[col].fillna(replace_with)
iris_imp[col].head()
```

```python
iris_imp.head()
```

```python
iris_imp.isna().sum()
```

```python
iris_imp[col].median()
```

```python
iris_imp[col].mode()[0]
```

nel caso della moda, spesso non viene restituito un solo valore ma una serie di valori, per questo si specifica con l'indice quale valore di moda considerare

```python
iris_imp.copy()
```

```python
replace_with = iris_imp.mean(numeric_only=True)
iris_imp = round(iris_imp.fillna(replace_with), 1)
iris_imp.head()
```

```python
iris_imp.isna().sum()
```

#### Si può utilizzare un trasformatore di SciKit-Learn

```python
from sklearn.impute import SimpleImputer
```

```python
X = iris.drop("species", axis=1).values
X.shape
```

```python
imp = SimpleImputer(strategy= "mean")
X_imp = imp.fit_transform(X)
X_imp[:5]
```
