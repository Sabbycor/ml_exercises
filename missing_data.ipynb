Metodo 1: Rimuovere righe o colonne con valori mancanti

```python
iris_drop = iris.copy()
```

```python
iris_drop = iris.dropna()
iris_drop.head()
```

```python
iris_drop.shape
```

```python
iris.shape
```

```python
iris_drop = iris_drop.dropna(axis=1)
iris_drop.head()
```

```python
iris_drop.shape
```

```python
tresh = int(iris.shape[0]*0.9)
tresh
```

<b> thresh </b> serve a stabilire una soglia minima di valori validi che una riga o una colonna deve avere per essere mantenuta.

Quindi:

- se lo usi sulle righe, tieni solo le righe che hanno almeno un certo numero di valori non mancanti
- se lo usi sulle colonne, tieni solo le colonne che hanno almeno un certo numero di valori non mancanti

tresh = int(iris.shape[0] * 0.9) serve a trasformare una percentuale in un numero assoluto di righe valide richieste.
- iris.shape[0] restituisce il numero totale di righe del dataset, che nell'esempio sono 150
- 150 * 0.9 = 135
- int(...) converte il risultato in intero, quindi 135

Quindi tresh = 135 significa:

voglio tenere solo le colonne che hanno almeno 135 valori non mancanti
oppure, detto in modo equivalente, voglio scartare le colonne che hanno troppi valori mancanti

Si fa questa operazione perché il parametro thresh di dropna() *non accetta una percentuale*, ma un numero minimo di valori validi.
Se tu vuoi dire “tieni solo le colonne con almeno il 90% di dati presenti”, devi prima convertire quel 90% in un numero, cioè 135 nel tuo caso.

```python
iris_drop = iris_drop.dropna(axis=1, thresh=145) #treshold = specifichiamo i valori minimi validi che devono essere in una colonna
iris_drop.head()
```

nel caso delle righe, basta non definire l'asse

```python
iris_drop.shape
```

se vogliamo rimuovere tutte le colonne e le righe che hanno valori mancanti:

```python
iris_drop = iris_drop.dropna(how="all")
```

se vogliamo solo determinate colonne:

```python
iris_drop = iris_drop.dropna(subset= ["petal_width"])
iris_drop.head(10)
```

in questo caso verranno rimosse le righe dove il valore della colonna selezionata non è disponibile.
